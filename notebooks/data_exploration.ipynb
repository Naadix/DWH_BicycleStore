```python
import pandas as pd 
```

## Data Exploration 

```python
# define the list of files 
file_list = ['brands.csv','categories.csv','customers.csv','order_items.csv','order_items.parquet','orders.csv','products.csv','staffs.csv','stocks.csv','stores.csv'] 
dataframes = {}

for file in file_list:
    print(f"=============== {file.upper()} ===============")
    file_path = f'../data/{file}'
    if file.endswith('.csv'):
        df = pd.read_csv(file_path)
    elif file.endswith('.parquet'):
        df = pd.read_parquet(file_path)
    else:
        print('THIS FILE IS NOT SUPPORTED')
        continue

    # store tables inside dictionary
    dataframes[file] = df

   # explore data 
    print(f'shape: {df.shape} ')
    print(f'head: \n {df.head()} ')
    print(f'types: \n {df.dtypes} ')
    print(f'info: \n {df.info()} ')
    print(f'describe: \n {df.describe()} ')
    print(f'total null values: \n {df.isnull().sum()}')
    print('\n','='*30, '\n')
```
